```python
import pandas as pd
import mysql.connector
```

```python
#Connecting MYSQL

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Password"
)
```

```python
#Create Cursor

cursor = connection.cursor()
print("Cursor created successfully!")
```

### What is a Cursor?

A cursor is an object used to execute SQL statements and retrieve results from a database.

In Python, the cursor acts as an interface between the Python program and the MySQL database.

The cursor allows us to:

- Execute SQL queries
- Create database objects
- Insert or modify data
- Retrieve query results

```python
cursor.execute("""
CREATE DATABASE IF NOT EXISTS retail_db
""")

print("Database created successfully!")
```

### CREATE DATABASE

The `CREATE DATABASE` statement creates a new database in MySQL.

Syntax:

CREATE DATABASE database_name;

The `IF NOT EXISTS` condition prevents an error if the database already exists.

Example:

CREATE DATABASE IF NOT EXISTS retail_db;

```python
cursor.execute("""
USE retail_db
""")

print("Using retail_db database")
```

### USE Statement

The `USE` statement selects the database that will be used for subsequent SQL operations.

Example:

USE retail_db;

After executing this statement, tables and other database objects will be created inside `retail_db`.

```python
cursor.execute("SHOW DATABASES")

databases = cursor.fetchall()

for database in databases:
    print(database[0])
```

### SHOW DATABASES

The `SHOW DATABASES` statement displays all databases available on the MySQL server.

Example:

SHOW DATABASES;

This can be used to verify that the `retail_db` database was successfully created.
