# Data Quality Assessment

## Objective

The purpose of this section is to evaluate the completeness, consistency, and uniqueness of the dataset before performing any cleaning operations.

Data quality directly affects the reliability of business insights. Therefore, missing values, duplicate records, and inconsistent entries must be identified before analysis.

---

## Why Data Quality Assessment Matters

Poor-quality data can lead to:

* Incorrect revenue calculations
* Misleading customer segmentation
* Inaccurate sales forecasts
* Wrong business recommendations

A systematic assessment helps determine which issues require cleaning and which represent valid business events such as cancelled orders.

## Observations – Missing Values

* The dataset is largely complete, with only two columns containing missing values.
* `Description` has 2,928 missing values (0.56%), which represents a very small proportion of the dataset.
* `Customer ID` has 107,927 missing values (20.54%), indicating that a substantial number of transactions cannot be linked to individual customers.
* Sales-related analyses can still use these transactions, but customer-level analyses such as RFM segmentation will require valid Customer IDs.

## Observations – Duplicate Records

* The dataset contains **6,865 duplicate rows**, representing approximately **1.31%** of all records.
* Exact duplicates are unlikely to represent additional customer purchases because every attribute is identical.
* These records will be removed during the data cleaning phase to prevent double-counting of revenue and transaction volume.

## Observations – Dataset Diversity

* The dataset contains **28,816 invoices**, indicating a large number of customer transactions.
* There are **4,632 unique products**, providing a broad product portfolio for sales and inventory analysis.
* The dataset includes **4,383 unique customers**, which is sufficient for customer segmentation and retention analysis.
* Customers belong to **40 different countries**, enabling international market performance analysis.
